### 43. Ek simple calculator banayein (+, -, *, /) symbols ke basis par operation perform karein.

```python
a = float(input("Pehla number: "))
op = input("Operator (+, -, *, /): ")
b = float(input("Doosra number: "))
if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print(a * b)
elif op == "/":
    print(a / b if b != 0 else "Division by zero nahi ho sakti")
```

### 58. Electricity Bill: Pehle 100 units free, next 100 units Rs 5/unit, uske baad Rs 10/unit.

```python
units = int(input("Units consumed: "))
if units <= 100:
    bill = 0
elif units <= 200:
    bill = (units - 100) * 5
else:
    bill = 100 * 5 + (units - 200) * 10
print("Bill:", bill, "Rs")
```

### 60. Final Challenge: Ek mini-project logic banayein jisme user se Username, Password, aur ek Security Question poochein nested loops aur if-else ke saath.

```python
correct_user = "ahmed"
correct_pass = "secret123"
security_answer = "karachi"

username = input("Username: ")
if username == correct_user:
    password = input("Password: ")
    if password == correct_pass:
        city = input("Security Question - Aapka shehar kya hai? ").lower()
        if city == security_answer:
            print("Login Successful! Welcome,", username)
        else:
            print("Security question wrong")
    else:
        print("Wrong Password")
else:
    print("Username not found")
```
